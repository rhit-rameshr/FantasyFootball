```python
import pandas as pd

# Define nflverse url - no season loaded yet

url = "https://github.com/nflverse/nflverse-data/releases/download/player_stats/player_stats.csv"
```

```python
df = pd.read_csv(url)

# Make sure season is numeric so comparisons work reliably

df["season"] = pd.to_numeric(df["season"], errors="coerce")

print(df.shape)
print(df["season"].unique()[:10])

# (rows, columns) = df.shape
```

```python
# Use the latest season available in the dataset instead of assuming 2025 exists

SEASON = int(df["season"].max())
df_2024 = df[df["season"] == SEASON]

print(f"Latest available season: {SEASON}")
print(df_2024.shape)
```

```python
player_df = df_2024[
    df_2024["player_display_name"] == "Christian McCaffrey"
]

player_df[
    [
        "player_display_name",
        "recent_team",
        "week",
        "opponent_team",
        "season_type",
        "fantasy_points",
        "fantasy_points_ppr"
    ]
]
```

```python
df_2024_reg = df_2024[
    df_2024["season_type"] == "REG"
].copy()

print("Before:", df_2024.shape)
print("After:", df_2024_reg.shape)

print(df_2024_reg["season_type"].value_counts())

# Filter to only regular season games since we don't need postseason
```

```python
duplicates = df_2024_reg.duplicated(
    subset=["player_id", "season", "week"]
)

print("Duplicate player-week rows:", duplicates.sum())

# Make sure there aren't any duplicates

# Make sure there are no missing important values
print("Missing player IDs:", df_2024_reg["player_id"].isna().sum())
print("Missing player names:", df_2024_reg["player_display_name"].isna().sum())
print("Missing positions:", df_2024_reg["position"].isna().sum())
print("Missing teams:", df_2024_reg["recent_team"].isna().sum())
print("Missing opponents:", df_2024_reg["opponent_team"].isna().sum())
```
